```python
# pip install -qU langchain langchain-ollama
from langchain.agents import create_agent

def get_weather(city: str) -> str:
    """Get weather for a given city."""
    return f"It's always sunny in {city}!"

agent = create_agent(
    model="ollama:qwen3.5:27b",
    tools=[get_weather],
    system_prompt="You are a helpful assistant",
)

result = agent.invoke(
    {"messages": [{"role": "user", "content": "What's the weather in San Francisco?"}]}
)
print(result["messages"][-1].content_blocks)
```

```python
from langchain.tools import tool
from langchain.agents import create_agent

def find_similar_query(query: str) -> list:
    """Find similar queries."""
    return [
        {
        "isFraud": True,
        "TransctionID": "1234567890",
        "id_01": -2
        },
        {
            "isFraud": True,
            "TransctionID": "4567890",
            "id_01": 50
        }
    ]

def read_shap_values(query: str) -> str:
    """Read SHAP values for a given query."""
    return f"SHAP values for {query}: [0.2, 0.3]"

# Create a subagent
find_similar_subagent = create_agent(model="ollama:qwen3.5:27b", tools=[find_similar_query])
read_shap_subagent = create_agent(model="ollama:qwen3.5:27b", tools=[read_shap_values])

# Wrap it as a tool
@tool("find_similar_fraud_queries", description="Find similar queries to justify why current query is fraudulent")
def call_similar_fraud_queries_agent(query: str):
    result = find_similar_subagent.invoke({"messages": [{"role": "user", "content": query}]})
    return result["messages"][-1].content

@tool("read_shap_values", description="Read SHAP values for a given query")
def call_read_shap_values_agent(query: str):
    result = read_shap_subagent.invoke({"messages": [{"role": "user", "content": query}]})
    return result["messages"][-1].content



# Main agent with subagent as a tool
main_agent = create_agent(model="ollama:qwen3.5:27b", tools=[call_similar_fraud_queries_agent, call_read_shap_values_agent])
```
